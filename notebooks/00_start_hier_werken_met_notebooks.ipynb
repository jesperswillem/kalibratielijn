## 1. Twee soorten cellen

Een notebook bestaat uit cellen.

**Tekstcellen** bevatten uitleg, formules en opdrachten. Deze cel is een tekstcel.

**Codecellen** bevatten Python-code. Die cellen kun je uitvoeren. De uitkomst verschijnt direct onder de cel.


## 2. Een cel uitvoeren

Klik op de codecel hieronder en druk daarna op:

- `Shift + Enter`, of
- de afspeelknop links van de cel.


```
print('Hallo! Deze codecel is uitgevoerd.')
```

## 4. Wat betekent `[*]`?

Soms zie je links van een codecel `[*]`. Dat betekent dat de cel bezig is met uitvoeren.

Als een cel klaar is, zie je meestal een nummer, bijvoorbeeld `[3]`.

## 6. De hele notebook opnieuw draaien

Als je opnieuw wilt beginnen:

**In Google Colab**

1. Kies `Runtime`.
2. Kies `Restart runtime`.
3. Kies daarna `Run all`.

**In JupyterLab of Jupyter Notebook**

1. Kies `Kernel`.
2. Kies `Restart Kernel and Run All Cells`.

Dit is handig als je per ongeluk cellen in een verkeerde volgorde hebt uitgevoerd.
